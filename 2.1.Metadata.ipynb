```python
import os
import json

from dotenv import load_dotenv
from openai import OpenAI

load_dotenv()
AI_GATEWAY_TOKEN = os.getenv("AI_GATEWAY_TOKEN")
CLOUDFLARE_ACCOUNT_ID = os.getenv("CLOUDFLARE_ACCOUNT_ID")

client = OpenAI(api_key=AI_GATEWAY_TOKEN, base_url=f"https://gateway.ai.cloudflare.com/v1/{CLOUDFLARE_ACCOUNT_ID}/ai-gateway-course/openai")

response = client.chat.completions.create(
    model="gpt-6-luna",
    messages=[
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": "넌 무슨 모델이지"},
    ],
    # 이 값은 쿼리에 사용할 수 있다.
    extra_headers={
        "cf-aig-metadata": json.dumps({"user_id":"1234567890","is_paid":True, "plan": "Pro"})
    }
)

print(response.choices[0].message.content)
```
