```python
import json
import os

from dotenv import load_dotenv
from openai import OpenAI, omit

load_dotenv()
AI_GATEWAY_TOKEN = os.getenv("AI_GATEWAY_TOKEN")
CLOUDFLARE_ACCOUNT_ID = os.getenv("CLOUDFLARE_ACCOUNT_ID")

# 공급자 Base URL에 /v1이 이미 있으면 아래 끝의 /v1을 제거한다.
client = OpenAI(
    api_key="unused",
    base_url=f"https://gateway.ai.cloudflare.com/v1/{CLOUDFLARE_ACCOUNT_ID}/ai-gateway-course/custom-illuwa/v1",
    default_headers={
        "cf-aig-authorization": f"Bearer {AI_GATEWAY_TOKEN}",
    }
)

response = client.chat.completions.create(
    model="gpt-6-luna",
    messages=[
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": "넌 무슨 모델이지"},
    ],
    # 이 값은 쿼리에 사용할 수 있다.
    extra_headers={
        "Authorization": omit,  # Gateway에 저장된 default 키 사용
        "cf-aig-metadata": json.dumps({"user_id":"1234567890","is_paid":True, "plan": "Pro"})
    }
)

print(response.choices[0].message.content)
```
